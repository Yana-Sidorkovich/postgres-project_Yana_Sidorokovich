```python
import pandas as pd
data = pd.read_csv("./data/adult.data.csv")
data.head(10)
```

```python
#1. Calculate how many men and women (attribute sex) are represented in this dataset.
data['sex'].value_counts()
```

```python
#2. What is the average age of men (attribute age) across the entire dataset?
avg_age_men = data[data['sex'] == 'Male']['age'].mean().round(1)

print(f"Average age of men: {avg_age_men:.1f}")
```

```python
#3. What is the proportion of United States citizens (native-country attribute)?
prop_us_citizen = (data['native-country'] == 'United-States').mean()
print(f"Proportion of United States citizens: {prop_us_citizen}")
```

```python
#4-5. Calculate the mean and standard deviation of the age of those who earn more than 50K per year (the salary attribute) and those who earn less than 50K per year
avg_age_greater50k = data[data['salary'] == '>50K']['age'].mean().round(1)
print(f"Average age of those who earn >50K: {avg_age_greater50k}")

avg_age_less50k = data[data['salary'] == '<=50K']['age'].mean().round(1)
print(f"Average age of those who earn <=50K: {avg_age_less50k}")

std_age_greater50k = data[data['salary'] == '>50K']['age'].std().round(1)
print(f"standard deviation of age for >50K: {std_age_greater50k}")

std_age_less50k = data[data['salary'] == '<=50K']['age'].std().round(1)
print(f"standard deviation of age for <=50K: {std_age_less50k}")
```

```python
#6. Is it true that people who earn more than 50k have at least a higher education? (education indicator – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate)
rich_people = data[data['salary'] == '>50K']

higher_ed_mask = rich_people['education'].str.contains('bachelors|prof-school|assoc-acdm|Assoc-voc|Masters|Doctorate', case=False, na=False)

is_true = higher_ed_mask.all()

print(f"Is it true that people who earn more than 50k have at least a higher education? {is_true}")
```

```python
#7. Find age statistics for each race (race attribute) and each gender. Use groupby and describe. Find the maximum age of males of the Asian-Pac-Islander race.
# Выбираем два столбца и применяем describe()
# Группируем сразу по расе и полу, считаем статистику по возрасту
stats = data.groupby(['race', 'sex'])['age'].describe().round(1)
max_age = stats.loc[('Asian-Pac-Islander', 'Male'), 'max']
print(stats)

print(f"\n\nMaximum age of males of the Asian-Pac-Islander race: {max_age}")
```

```python
#8. Among whom is the proportion of high-earning individuals (>50K) higher: married or single men (marital-status indicator)? We consider those whose marital-status begins with C (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse) to be married; the rest are considered single.
is_married_men = ((data['sex'] == 'Male') &
                  (data['marital-status'].str.startswith('Married', na=False)))
is_single_men = ((data['sex'] == 'Male') & (~data['marital-status'].str.startswith('Married', na=False)))
married_men = data[is_married_men]
single_men = data[is_single_men]
rich_married = (married_men['salary'] == '>50K').mean()
rich_single = (single_men['salary'] == '>50K').mean()
if rich_married > rich_single:
    print("Among married men, the proportion of high earners is higher.")
else:
    print("Among single men, the proportion of high earners is higher.")
```

```python
#9. What is the maximum number of hours a person works per week (hours-per-week indicator)? How many people work that many hours, and what percentage of them earn a lot?
max_hours = data['hours-per-week'].max()
hardworkers_df = data[data['hours-per-week'] == max_hours]
hardworkers_count = len(hardworkers_df)
hardworkers_rich = (hardworkers_df['salary'] == '>50K').mean()*100

print(f"Maximum number of hours: {max_hours}")
print(f"The number of people working that many hours: {hardworkers_count}")
print(f"Percentage of them who earn a lot: {hardworkers_rich:.1f}%")
```

```python
#10. Calculate the average working hours (hours-per-week) of low and high earners (salary) for each country (native-country).
data['native-country'] = data['native-country'].replace('?', 'Unknown')
avg_hours = data.groupby(['native-country', 'salary'])['hours-per-week'].mean().round(1).unstack()
print(avg_hours)
```

```python
'''11.Group people into age groups young, adult, retiree, where:
young corresponds to 16-35 years old
adult - 35-70 years old
retiree - 70-100 years old
Enter the name of the appropriate age group for each person in the new AgeGroup column.'''
data['AgeGroup'] = ''

data.loc[(data['age'] >= 16) & (data['age'] < 35), 'AgeGroup'] = 'young'
data.loc[(data['age'] >= 35) & (data['age'] < 70), 'AgeGroup'] = 'adult'
data.loc[(data['age'] >= 70) & (data['age'] <= 100), 'AgeGroup'] = 'retiree'

check_groups = data.groupby('AgeGroup')['age'].agg(['count', 'min', 'mean', 'max']).round(1)
check_groups = check_groups.drop('', errors='ignore')
check_groups = check_groups.reindex(['young', 'adult', 'retiree'])
print("Verification of created age groups:")
print(check_groups)
```

```python
#12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people most often earn more than 50K (>50K)
rich_people = data[data['salary'] == '>50K']
rich_by_age = rich_people['AgeGroup'].value_counts()

print(rich_by_age)
top_group = rich_by_age.sort_values(ascending=False).index[0]
print(f"\n\nThe richect group: {top_group}")
```

```python
#14. Group people by employment type (the occupation column) and determine the number of people in each group. Then write a filter function, filter_func, that will return only those groups in which the average age (the age column) is 40 or less and in which all employees work more than 5 hours per week (the hours-per-week column).
data['occupation'] = data['occupation'].replace('?', 'Not given')
emp_type = data.groupby('occupation').size()
print(emp_type)

result = data.groupby('occupation').filter(
    lambda x: (x['hours-per-week'] > 5).all() and (x['age'].mean() <= 40)
)

summary = result.groupby('occupation').agg(count=('age', 'size'), avg_age=('age', 'mean'), avg_hours=('hours-per-week', 'mean')).round(1)
print("Occupation groups after filtering.")
print(summary)
```
