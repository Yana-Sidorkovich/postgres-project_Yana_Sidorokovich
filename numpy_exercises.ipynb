```python
# 1. Given a random array, change the sign of elements whose values between 3 and 8
import numpy as np

# 1. Генерируем массив
arr = np.random.randint(0, 16, size=15)
print("1. Исходный массив:      ", arr)

# 2. Создаем маску для чисел между 3 и 8
mask = (arr > 3) & (arr < 8)

# 3. Меняем знак у элементов, которые попадают под маску
arr[mask] = -arr[mask]

print("2. После изменения знака:", arr)
```

```python
# 2. Replace the maximum element of a random array with 0

# 1. Создаем массив
arr2 = np.random.randint(0, 20, size=10)
print("1. Исходный массив:     ", arr2)

# 2. Находим максимум
max_value = arr2.max()
print("   Максимальный элемент:", max_value)

# 3. Заменяем на 0 (способ через маску)
arr2[arr2 == max_value] = 0

print("2. После замены:        ", arr2)
```

```python
# 4. Given two arrays, A (8x3) and B (2x2), find rows in A that contain elements from each row in B, regardless of the order of the elements in B.
# Создаем тестовые массивы
A = np.array([
    [1, 2, 3],
    [4, 5, 6],
    [2, 1, 7],  # содержит 1 и 2 (из первой строки B)
    [8, 9, 10],
    [5, 4, 11], # содержит 4 и 5 (из второй строки B)
    [1, 3, 5],
    [6, 4, 5],  # содержит 4 и 5
    [2, 8, 9]
])

B = np.array([
    [1, 2],  # ищем строки, содержащие 1 и 2
    [4, 5]   # ищем строки, содержащие 4 и 5
])

# Решение: проверяем каждую строку A
result_rows = []

for row_b in B:  # для каждой строки в B
    for i, row_a in enumerate(A):  # для каждой строки в A
        # Проверяем, все ли элементы из row_b есть в row_a
        if np.all(np.isin(row_b, row_a)):
            result_rows.append(i)

# Убираем дубликаты индексов
result_rows = list(set(result_rows))
result = A[result_rows]

print("Строки из A, содержащие элементы из B:")
print(result)
print("Индексы строк:", result_rows)
```

```python
# 5. Given a 10x3 matrix, find rows of unequal values (for example, the row [2,2,3] remains, the row [3,3,3] is deleted)
# Создаем тестовую матрицу 10x3
matrix = np.array([
    [1, 2, 3],  # остается (разные)
    [2, 2, 2],  # удаляется (все равны)
    [4, 5, 4],  # остается
    [3, 3, 3],  # удаляется
    [1, 1, 2],  # остается
    [7, 7, 7],  # удаляется
    [8, 9, 10], # остается
    [5, 5, 5],  # удаляется
    [1, 2, 1],  # остается
    [6, 6, 6]   # удаляется
])

# Способ 1: Проверяем, все ли элементы в строке равны первому
mask = ~np.all(matrix == matrix[:, [0]], axis=1)
result = matrix[mask]

print("Строки с неравными значениями:")
print(result)

# Способ 2: Через min и max (более понятный)
mask2 = (matrix.min(axis=1) != matrix.max(axis=1))
result2 = matrix[mask2]

print("\nТот же результат другим способом:")
print(result2)
```

```python
#For each of the following problems (1-5) you need to provide 2 implementations – one without using numpy (Consider that where the input or output should be a numpy array, there will simply be lists), and the second one is fully vectorized using numpy (without using Python loops/map/list comprehension).

#Note 1. It can be assumed that all specified objects are non-empty (for example, in task 1 there are non-zero elements on the diagonal of the matrix).

#Note 2. For most tasks, the solution takes no more than 1-2 lines.

#Task 6.1: Calculate the product of nonzero elements on the diagonal of a rectangular matrix.
#For example, for X = np.array([[1, 0, 1], [2, 0, 2], [3, 0, 3], [4, 4, 4]]) answer is 3.

import numpy as np
from math import prod

X = np.array([[1, 0, 1],
              [2, 0, 2],
              [3, 0, 3],
              [4, 4, 4]])

#  Решение БЕЗ NumPy
def diagonal_product_no_numpy(matrix):
    # Получаем диагональ (минимальное из rows/cols)
    rows = len(matrix)
    cols = len(matrix[0]) if rows > 0 else 0
    diag_length = min(rows, cols)

    # Извлекаем диагональ
    diagonal = [matrix[i][i] for i in range(diag_length)]

    # Фильтруем нули и считаем произведение
    nonzero_diag = [x for x in diagonal if x != 0]
    return prod(nonzero_diag) if nonzero_diag else 1

# ✅ Решение С NumPy
def diagonal_product_numpy(matrix):
    diag = np.diag(matrix)  # извлекаем диагональ
    return np.prod(diag[diag != 0])  # произведение ненулевых

print("Без NumPy:", diagonal_product_no_numpy(X.tolist()))
print("С NumPy:", diagonal_product_numpy(X))
```

```python
#Task 6.2: Two vectors x and y are given. Check if they define the same multiset.
#For example, for x = np.array([1, 2, 2, 4]), y = np.array([4, 2, 1, 2]) answer is True.

import numpy as np

x = np.array([1, 2, 2, 4])
y = np.array([4, 2, 1, 2])

#  Решение БЕЗ NumPy
def same_multiset_no_numpy(x, y):
    return sorted(x) == sorted(y)

# ✅ Решение С NumPy
def same_multiset_numpy(x, y):
    return np.array_equal(np.sort(x), np.sort(y))

print("Без NumPy:", same_multiset_no_numpy(x.tolist(), y.tolist()))
print("С NumPy:", same_multiset_numpy(x, y))
```

```python
#Task 6.3: Find the maximum element in the vector x among the elements preceded by zero. For example, for x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0]) answer is 5.

import numpy as np

x = np.array([6, 2, 0, 3, 0, 0, 5, 7, 0])
# Индексы:    0  1  2  3  4  5  6  7  8
# После 0 идут:    3  0  5  (элементы на позициях 3, 5, 7)

#  Решение БЕЗ NumPy
def max_after_zero_no_numpy(x):
    result = []
    for i in range(len(x) - 1):
        if x[i] == 0:
            result.append(x[i + 1])
    return max(result) if result else None

# ✅ Решение С NumPy
def max_after_zero_numpy(x):
    # Находим индексы, где элемент = 0
    zero_indices = np.where(x[:-1] == 0)[0]
    # Берем элементы ПОСЛЕ каждого нуля
    elements_after_zero = x[zero_indices + 1]
    return np.max(elements_after_zero)

print("Без NumPy:", max_after_zero_no_numpy(x.tolist()))
print("С NumPy:", max_after_zero_numpy(x))
```

```python
#Task 6.4: Implement run-length encoding. For some vector x, it is necessary to return a tuple of two vectors of the same length. The first one contains numbers and the second one contains how many times they need to be repeated.
#For example, for x = np.array([2, 2, 2, 3, 3, 3, 5]) answer is (np.array([2, 3, 5]), np.array([3, 3, 1])).

import numpy as np

x = np.array([2, 2, 2, 3, 3, 3, 5])
# Ответ: (values=[2, 3, 5], counts=[3, 3, 1])

# ❌ Решение БЕЗ NumPy
def rle_no_numpy(x):
    if not x:
        return [], []

    values = [x[0]]
    counts = [1]

    for i in range(1, len(x)):
        if x[i] == x[i-1]:
            counts[-1] += 1
        else:
            values.append(x[i])
            counts.append(1)

    return values, counts

# ✅ Решение С NumPy
def rle_numpy(x):
    # Находим позиции, где элемент меняется
    diff_indices = np.where(x[:-1] != x[1:])[0] + 1
    # Добавляем начало и конец
    indices = np.concatenate(([0], diff_indices, [len(x)]))
    # Извлекаем уникальные значения
    values = x[indices[:-1]]
    # Считаем длины серий
    counts = np.diff(indices)
    return values, counts

values, counts = rle_numpy(x)
print("Values:", values)   # [2 3 5]
print("Counts:", counts)   # [3 3 1]
```

```python
# Task 6.5: Two samples of objects are given - X and Y. Calculate the matrix of Euclidean distances between objects. Compare with function scipy.spatial.distance.cdist by operating speed.

import numpy as np
from scipy.spatial.distance import cdist
import time

# Тестовые данные
X = np.array([[1, 2], [3, 4], [5, 6]])
Y = np.array([[0, 1], [2, 3]])

# ❌ Решение БЕЗ NumPy (наивное)
def euclidean_distances_no_numpy(X, Y):
    result = []
    for x in X:
        row = []
        for y in Y:
            # Евклидово расстояние: sqrt(sum((xi - yi)^2))
            dist = sum((xi - yi) ** 2 for xi, yi in zip(x, y)) ** 0.5
            row.append(dist)
        result.append(row)
    return result

# ✅ Решение С NumPy (векторизованное)
def euclidean_distances_numpy(X, Y):
    # Используем broadcasting
    # X: (n, d) -> (n, 1, d)
    # Y: (m, d) -> (1, m, d)
    # Разность: (n, m, d)
    diff = X[:, np.newaxis, :] - Y[np.newaxis, :, :]
    # Квадрат разности и сумма по последнему измерению
    distances = np.sqrt(np.sum(diff ** 2, axis=2))
    return distances

# ✅ Решение с scipy (для сравнения скорости)
def euclidean_distances_scipy(X, Y):
    return cdist(X, Y, metric='euclidean')

# Проверяем результаты
print("NumPy result:\n", euclidean_distances_numpy(X, Y))
print("SciPy result:\n", euclidean_distances_scipy(X, Y))

# Сравнение скорости
n_runs = 1000

# Замер NumPy
start = time.time()
for _ in range(n_runs):
    euclidean_distances_numpy(X, Y)
numpy_time = time.time() - start

# Замер SciPy
start = time.time()
for _ in range(n_runs):
    cdist(X, Y, metric='euclidean')
scipy_time = time.time() - start

print(f"\nNumPy время: {numpy_time:.4f} сек")
print(f"SciPy время: {scipy_time:.4f} сек")
print(f"SciPy быстрее в {numpy_time/scipy_time:.2f} раз")
```

```python
#Task 6: CrunchieMunchies
import numpy as np

calorie_stats = np.loadtxt("./data/cereal.csv", delimiter=",")
calorie_stats
```

```python
#6.2 One serving of CrunchieMunchies contains 60 calories. How much higher is the average calorie count of your competitors?

# Task 6.2
average_calories = np.mean(calorie_stats) - 60
print("Разница со средней калорийностью конкурентов:", average_calories)
```

```python
#6.3 Does the average calorie count accurately reflect the distribution of the data set? Let's sort the data and see.
# Task 6.3
calorie_stats_sorted = np.sort(calorie_stats)
print("Отсортированные данные:")
print(calorie_stats_sorted)
```

```python
#6.4
#It looks like most values are above the mean. Let's see if the median is the most accurate measure of the data set.

# Task 6.4
median_calories = np.median(calorie_stats)
print("Медиана:", median_calories)
print("Среднее значение:", np.mean(calorie_stats))
```

```python
#6.5
#While the median shows that at least half of our values are over 100 calories, it would be more impressive to show that a significant portion of the competitors have higher calorie counts than CrunchieMunchies.

# Task 6.5
# Проверяем разные процентили
for percentile in range(0, 100, 5):
    value = np.percentile(calorie_stats, percentile)
    print(f"{percentile}-й процентиль: {value}")
    if value > 60:
        nth_percentile = percentile
        break

print(f"\nМинимальный процентиль, превышающий 60 калорий: {nth_percentile}")
```

```python
#6.6
#While the percentile shows us that most competitors have much higher calorie counts, it's an awkward concept to use in marketing materials.

# Task 6.6
more_calories = np.mean(calorie_stats > 60) * 100
print(f"Процент хлопьев с калорийностью выше 60: {more_calories}%")
```

```python
#6.7
#This is a really high percentage. This will be very useful as we promote CrunchieMunchies. But one question is: how big are the differences in the dataset? Can we generalize that most cereals contain around 100 calories or is the range even wider?

# Task 6.7
calorie_std = np.std(calorie_stats)
print("Стандартное отклонение:", calorie_std)

# Дополнительно: можно показать диапазон
mean = np.mean(calorie_stats)
print(f"\nСреднее: {mean}")
print(f"Диапаон [среднее ± std]: [{mean - calorie_std}, {mean + calorie_std}]")
```
